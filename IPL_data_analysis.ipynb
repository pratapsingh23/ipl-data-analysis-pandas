```python
import pandas as pd

df = pd.read_csv('matches.csv')

df.head(15)
```

```python
# Checking missing values:
df.isnull().sum()
```

```python
# filling blanks in city according to venue:

df['city'] = df['city'].fillna(df['venue'])

```

```python
# filling blanks in winner cloumn with string "No Result

df['winner'].fillna('No Result')

```

```python
# removing 'umpire 3' column:
df.drop(columns = ['umpire3'], inplace = True)
```

```python
# Checking columns:

print(df.columns)
```

```python
# Finding Total matches:
total_matches = len(df['id'])
print("Total Matches", total_matches)
```

```python
# How many teams played?

df['team1'].unique()
```

```python
# which team has won most of matches?

print("Most win of each teams: \n")
df['winner'].value_counts()
```

```python
# Highest number of Player of the match:

print("Top 5 POM players: ")
df['player_of_match'].value_counts().head(5).reset_index()
```

```python
# Toss Winner vs Match Winner Analysis:
sum(df['toss_winner'] == df['winner'])
```

```python

# Toss Decision Breakdown:

toss_break = df['toss_decision'].value_counts().reset_index()
display(toss_break)

```

```python
# Most matches played in stadiums:
df['venue'].value_counts().head(5)
```

```python
# Batting first vs chasing win analysis:

# bat first:
bat_first_win = len(df[df['win_by_runs'] > 0])
bat_first_pct = (bat_first_win/total_matches)*100
print(f"Batting first wins: {bat_first_win}\nPercentage: {bat_first_pct:.2f}%\n")

# bowl first:
bowl_first_win = len(df[df['win_by_wickets'] > 0])
bowl_first_pct = (bowl_first_win/total_matches)*100
print(f"Win by chase: {bowl_first_win}\nPercnetage: {bowl_first_pct:.2f}%")

```

```python
# season wise total matches:
season_wise = df.groupby('season')['id'].count().reset_index()
display(season_wise)
```

```python
#In top venue : Batting or Bowling:

chinnaswami = df[df['venue'] == 'M Chinnaswamy Stadium']
chasing_chinnaswami = len(chinnaswami[chinnaswami['win_by_wickets'] > 0])
print("Chinnaswami chasing wins: ",chasing_chinnaswami)

batting_chinnaswami = len(chinnaswami[chinnaswami['win_by_runs'] > 0])
print("Chinnaswami batting wins: ", batting_chinnaswami)
```

```python
# Final Winner of every IPL season:
final_winner = df.groupby('season')['winner'].last().reset_index()
display(final_winner)
```

```python
# Win by biggest runs margin:

df[df['win_by_runs'] == df['win_by_runs'].max()][['season', 'team1', 'team2', 'winner', 'win_by_runs']]
```

```python
# Win by biggest wickets margin:

df[df['win_by_wickets'] == df['win_by_wickets'].max()][['season', 'team1', 'team2', 'winner', 'win_by_wickets']]
```

```python
# Final Task: MI vs CSK:

# filtering:

mi_csk = df[((df['team1'] == 'Mumbai Indians') & (df['team2'] == 'Chennai Super Kings')) |
            ((df['team1'] == 'Chennai Super Kings') & (df['team2'] == 'Mumbai Indians'))]

# Head to head win count

mi_csk['winner'].value_counts().reset_index()
```
